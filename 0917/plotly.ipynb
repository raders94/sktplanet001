```python
import pandas as pd
import plotly.express as px

df= pd.DataFrame({
    'items': ['노트북','모니터','키보드','마우스'],
    'sales': [12,5,7,9],

})
df
```

```python
fig=px.bar(
        df,
        x='items',
        y='sales',
        title='제품별 매출'
    )

fig.show()
```

```python
dx= pd.DataFrame({
        'month': ['jan','feb','mar','apr','may','jun'],
        'sales': [120,50,70,90,160,180],

    })

fig =px.line(dx,
        x='month',
        y='sales',
        markers=True,
        title='월별 매출 변화량')
fig.show()
```

```python
df= pd.DataFrame({
        'month': ['jan','jan','feb','feb','mar','mar'],
        'category': [
            'pc','주변기기',
            'pc','주변기기',
            'pc','주변기기',
                     ],
        'sales': [ 120,80,75,46,13,49,]

    })

fig= px.bar(df,
            x='month',
            y='sales',
            color='category',
            barmode='group',
            title='월별 카테고리 매출')

fig.show()
```
